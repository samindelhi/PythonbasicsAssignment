**2. What is an interpreter in Python?**
  * In Python, an interpreter is a program that executes Python code line by line, converting it into machine-readable instructions. Unlike compiled languages (like C or Java), which translate the entire code into machine code before running it, Python executes each instruction immediately.

**3. What are pre-defined keywords in Python**
  *  In Python, pre-defined keywords (or reserved words) are words that have special meanings in the language and cannot be used as variable names, function names, or identifiers. These keywords define Python's syntax and structure. Some common keywords are if, else, elif, for, while etc.

**4. Can keywords be used as variable names**
  * No, Python keywords cannot be used as variable names because they are reserved words that have predefined meanings in the language. If you try to use a keyword as a variable name, Python will throw a SyntaxError.

**5. What is mutability in Python?**
  * In Python, mutability refers to whether an object's value can be changed after it has been created viz. lists. Where the value cannot be changed after creation, that object is Immutable in nature viz. tuples.

**6. Why are lists mutable, but tuples are immutable?**
  * Lists are mutable due its Dynamic Memory Allocation. They are stored as referece to objects due to which their element can be modified without creating new list objects.
  Tuples are immmutable as they have Static or Fixed memory Allocation meaning, they store the value directly in the location instead of referencing which makes it difficult to change after creation / allocation.

  * "is" opearator checks for the stored memory location of the objects being checked. It doesn't check for the value but only the reference/identity of two objects. eg: a is b = false even when both a and b have same value.

````python
#6. Write a function to demonstrate the behavior of mutable and immutable arguments.

def modify_values(num, lst):
    num += 10  # Modifying an immutable integer (Creates a new object)
    lst.append(4)  # Modifying a mutable list (Changes the original list)

    print("Inside function:")
    print("num:", num)  # New value (but original remains unchanged)
    print("lst:", lst)  # Modified list

# Initialize immutable and mutable objects
num_val = 5  # Immutable (int)
list_val = [1, 2, 3]  # Mutable (list)

# Call function
modify_values(num_val, list_val)

print("\nOutside function:")
print("num_val:", num_val)  # Still 5 (Immutable, unchanged)
print("list_val:", list_val)  # Modified (Mutable, changed)
````

````python
#10. Write code to demonstrate type casting with list elements.
# Original list with mixed data types
data = ["10", "20", "30", "40"]

# Type casting: Convert string elements to integers
int_data = [int(i) for i in data]
print("Integer List:", int_data)  # Output: [10, 20, 30, 40]

# Type casting: Convert integers to floats
float_data = [float(i) for i in int_data]
print("Float List:", float_data)  # Output: [10.0, 20.0, 30.0, 40.0]

# Type casting: Convert integers to strings again
str_data = [str(i) for i in int_data]
print("String List:", str_data)  # Output: ['10', '20', '30', '40']
````
